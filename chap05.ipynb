Tokenizing datasets - loads the appropirate tokenizer class for the specified model

```python
from transformers import AutoTokenizer
from huggingface_hub import list_datasets
from datasets import load_dataset

dataset_id = 'stanfordnlp/imdb'

dataset = load_dataset(dataset_id)  

tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
tokenized_dataset = dataset.map(
    lambda examples: 
        tokenizer(examples['text'], 
                  truncation=True, 
                  padding='max_length'), 
    batched=True)
```

```python
print(tokenized_dataset)
```

```python
print(tokenized_dataset['train'][0]['input_ids'])
```

Convert the token ids back to tokens

```python
tokens = tokenizer.convert_ids_to_tokens(
             tokenized_dataset['train'][0]['input_ids'])
print(tokens)
```

```python
print(tokenized_dataset['train'][0]['attention_mask'])
```

Load the twitter financial news topic dataset

https://huggingface.co/datasets/zeroshot/twitter-financial-news-topic

```python
from datasets import load_dataset

# Step 1: Load the dataset
dataset = load_dataset('zeroshot/twitter-financial-news-topic')

# Step 2: Extract the topic labels from the training set
train_data = dataset['train']
```

```python
print(train_data[-1])
```

20 Labels used in the dataset

```python
topics = {                                    
    "LABEL_0": "Analyst Update",
    "LABEL_1": "Fed | Central Banks",
    "LABEL_2": "Company | Product News",
    "LABEL_3": "Treasuries | Corporate Debt",
    "LABEL_4": "Dividend",
    "LABEL_5": "Earnings",
    "LABEL_6": "Energy | Oil",
    "LABEL_7": "Financials",
    "LABEL_8": "Currencies",
    "LABEL_9": "General News | Opinion",
    "LABEL_10": "Gold | Metals | Materials",
    "LABEL_11": "IPO",
    "LABEL_12": "Legal | Regulation",
    "LABEL_13": "M&A | Investments",
    "LABEL_14": "Macro",
    "LABEL_15": "Markets",
    "LABEL_16": "Politics",
    "LABEL_17": "Personnel Change",
    "LABEL_18": "Stock Commentary",
    "LABEL_19": "Stock Movement",
}
```

```python
mapped_labels = [topics[f"LABEL_{label}"]     
                 for label in train_data['label']]
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Step 3: Plot the distribution of topics
plt.figure(figsize=(10, 6))

# Create histogram
bins = np.arange(len(topics) + 1) - 0.5       
plt.hist(mapped_labels, 
         bins = bins, 
         edgecolor = 'black', 
         color = 'skyblue', 
         alpha = 0.7)

plt.xticks(np.arange(len(topics)),            
           list(topics.values()), 
           rotation = 90, 
           ha = 'center')

plt.title("Topic Distribution - Twitter Financial News")
plt.xlabel("Topics")
plt.ylabel("Number of Tweets")
plt.tight_layout()
plt.show()
```

```python
from datasets import load_dataset
import matplotlib.pyplot as plt
import numpy as np

# Load the CIFAR-10 dataset
dataset = load_dataset('uoft-cs/cifar10')
print(dataset)
```

```python
dataset['train'][0]
```

```python
# Define labels for the CIFAR-10 classes
labels = {
    0: "airplane",
    1: "automobile",
    2: "bird",
    3: "cat",
    4: "deer",
    5: "dog",
    6: "frog",
    7: "horse",
    8: "ship",
    9: "truck"
}

# Visualize a few images from the training set
def show_images(images, labels, labels_dict):
    plt.figure(figsize=(5, 5))
    for i in range(25):
        plt.subplot(5, 5, i + 1)
        plt.imshow(images[i])
        plt.title(labels_dict[labels[i]])                            
        plt.axis('off')
    plt.tight_layout()
    plt.show()

# Get some samples from the training dataset
train_samples = dataset['train'].shuffle(seed=42).select(range(25))  

# Extract images and labels
images = [sample['img'] for sample in train_samples]
class_labels = [sample['label'] for sample in train_samples]

# Display the images with class names
show_images(images, class_labels, labels)
```
